# Day 1 & Day 2: Data Structures

```python
# Task 3.1: Student Names Loop
students = ["Ali", "Yaseen", "Ahmed", "Fatima", "Zainab"]
print("--- Student List ---")
for student in students:
    print(student)

# Task 3.2: Marks Analysis (Max, Min, Average)
marks = [78, 56, 89, 92, 65, 73]
max_mark = max(marks)
min_mark = min(marks)
avg_mark = sum(marks) / len(marks)

print("\n--- Marks Analysis ---")
print(f"Maximum Marks: {max_mark}")
print(f"Minimum Marks: {min_mark}")
print(f"Average Marks: {avg_mark:.2f}")

# Task 3.3: Word Frequency Counter
sentence = "I love Python because Python is easy"
words = sentence.split()
word_count = {}

for word in words:
    word_count[word] = word_count.get(word, 0) + 1

print("\n--- Word Frequency ---")
for word, count in word_count.items():
    print(f"'{word}': {count}")
```

# Day 3: Functions

```python
# Task 4.1: Factorial Function
def calculate_factorial(n):
    if n < 0:
        return "Factorial does not exist for negative numbers."
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

# Task 4.2: Prime Checker Function
def check_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Task 4.3: Reverse String Function (Without [::-1])
def reverse_text(text):
    reversed_str = ""
    for char in text:
        reversed_str = char + reversed_str
    return reversed_str

# Executing Function Tests
print("Factorial of 5:", calculate_factorial(5))
print("Is 17 Prime?:", check_prime(17))
print("Is 20 Prime?:", check_prime(20))
print("Reversed 'Artificial Intelligence':", reverse_text("Artificial Intelligence"))
```

# Day 4 & Day 5: Breadth-First Search

```python
# Task 5.1: 8-Node BFS Implementation
# Representing graph using Adjacency List
graph_8_nodes = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F", "G"],
    "D": ["H"],
    "E": [],
    "F": [],
    "G": [],
    "H": []
}

def bfs_traversal(graph, start_node):
    visited = []
    queue = [start_node]

    while queue:
        node = queue.pop(0)  # Dequeue first node (FIFO)
        if node not in visited:
            visited.append(node)
            for neighbor in graph.get(node, []):
                if neighbor not in visited:
                    queue.append(neighbor)
    return visited

print("BFS 8-Node Traversal Order:", bfs_traversal(graph_8_nodes, "A"))

# Task 5.2: Path Existence Check via BFS
def bfs_path_exists(graph, start, goal):
    visited = set()
    queue = [start]

    while queue:
        node = queue.pop(0)
        if node == goal:
            return True
        if node not in visited:
            visited.add(node)
            for neighbor in graph.get(node, []):
                if neighbor not in visited:
                    queue.append(neighbor)
    return False

print("Does path exist from 'A' to 'H'?:", bfs_path_exists(graph_8_nodes, "A", "H"))
print("Does path exist from 'E' to 'C'?:", bfs_path_exists(graph_8_nodes, "E", "C"))
```
