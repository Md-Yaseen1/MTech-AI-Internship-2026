# Tasks 2.1, 2.2, & 2.3

```python
# Task 2.1: Prime Number Checker
num = int(input("Enter a number to check prime status: "))
if num > 1:
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    print(f"{num} is {'a Prime' if is_prime else 'not a Prime'} number.")
else:
    print(f"{num} is not a Prime number.")

# Task 2.2: Multiplication Table
table_num = int(input("\nEnter a number for multiplication table: "))
for i in range(1, 11):
    print(f"{table_num} x {i} = {table_num * i}")

# Task 2.3: Cumulative Sum Loop
total_sum = 0
print("\nEnter numbers to add (enter 0 to stop):")
while True:
    val = float(input("Enter number: "))
    if val == 0:
        break
    total_sum += val
print(f"Total Cumulative Sum: {total_sum}")
```

# Student Report Card & Eligibility System

```python
# Week 1 Mini Project: Interactive Student Evaluation System
def student_evaluator():
    print("=== Week 1 Mini Project: Student Evaluator ===")
    student_name = input("Enter Student Name: ")
    num_subjects = int(input("Enter total number of subjects: "))

    scores = []
    for i in range(1, num_subjects + 1):
        score = float(input(f"Enter marks for Subject {i} (out of 100): "))
        scores.append(score)

    total_marks = sum(scores)
    percentage = (total_marks / (num_subjects * 100)) * 100

    if percentage >= 80:
        grade = "A (Passed with Distinction)"
    elif percentage >= 60:
        grade = "B (Passed)"
    elif percentage >= 50:
        grade = "C (Satisfactory)"
    else:
        grade = "F (Needs Improvement)"

    print("\n--- PERFORMANCE SUMMARY ---")
    print(f"Student: {student_name}")
    print(f"Total Marks Obtained: {total_marks} / {num_subjects * 100}")
    print(f"Percentage: {percentage:.2f}%")
    print(f"Final Grade: {grade}")


student_evaluator()
```
